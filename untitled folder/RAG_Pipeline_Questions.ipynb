## 1. What is RAG (Retrieval-Augmented Generation)?

A Large Language Model (LLM) only "knows" what it saw during training — it has no access to your private documents, and it can be confidently wrong (**hallucinate**) about facts it was never trained on.

**RAG** fixes this by giving the LLM a look-up step before it answers:

1. **Ingest** — collect your documents (PDFs, docs, web pages, slides, etc.)
2. **Embed** — convert each document/chunk into a numeric vector that captures its *meaning*
3. **Store** — save those vectors in a **vector database** for fast search
4. **Retrieve** — when a user asks a question, embed the question too, and fetch the most similar document vectors
5. **Rerank** *(optional but recommended)* — reorder the retrieved documents by true relevance using a more precise model
6. **Generate** — hand the retrieved text to the LLM as *context*, and ask it to answer **using only that context**

## 2. What are Embeddings?

An **embedding** is a list of numbers (a *vector*) that represents the meaning of a piece of text. Texts with similar meaning end up as vectors that are close together in that numeric space — even if they don't share any of the same words.

For example, *"The cat sat on the mat"* and *"A feline rested on the rug"* use completely different words but would produce **nearby** embedding vectors, because an embedding model captures semantics, not just spelling.

In this notebook we use Cohere's `embed-v4.0` model, which turns any string into a high-dimensional vector (e.g. 1024 or 1536 numbers, depending on the model).

**Why this matters for RAG:** once every document is a vector, "find the most relevant document for this question" becomes a pure **math problem** — measuring the distance between the question's vector and every document's vector.

## 3. What is a Vector Database (and what is FAISS)?

A **vector database** stores embeddings and lets you search them efficiently by *similarity* rather than by exact keyword match. Instead of `WHERE text = '...'`, a vector database answers questions like *"give me the 3 vectors closest to this one."*

**FAISS** (Facebook AI Similarity Search) is an open-source library from Meta AI that does exactly this — it is not a full managed database server, but a very fast **similarity-search engine** that can be embedded directly inside a Python program, which makes it ideal for learning and for small/medium projects. (For production systems at scale, managed vector databases such as Pinecone, Weaviate, Milvus, or Azure AI Search offer the same core idea with added persistence, filtering, and scaling.)

Two ideas you'll use directly in this notebook:

- **`IndexFlatL2`** — the simplest FAISS index. It computes the exact **L2 (Euclidean) distance** between the query vector and *every* stored vector, guaranteeing the true nearest neighbours (at the cost of being slower on huge datasets — fine for our small demo corpus).
- **`k` (top-k)** — the number of nearest neighbours to return. Asking for `k=3` means "give me the 3 most similar documents."

**Distance vs. similarity:** FAISS's `IndexFlatL2` returns a *distance* — **smaller is more similar** (0 = identical vectors), which is the opposite of a similarity *score*, where higher is better. Keep this in mind when you interpret results later in this notebook.

## 9. Reranking — Improving Precision

Vector search (Question 10) is fast but approximate about *relevance* — it only measures geometric closeness, not deep semantic relevance to the specific query. A **reranker** is a second, more expensive model that looks at the query and *each* candidate document **together**, and produces a much more accurate relevance ranking.

This two-stage pattern — cheap, fast **retrieval** over the whole corpus, followed by an expensive, accurate **rerank** over just the short-list — is standard practice in production RAG systems, because running the expensive model over *every* document in a large corpus would be far too slow.
